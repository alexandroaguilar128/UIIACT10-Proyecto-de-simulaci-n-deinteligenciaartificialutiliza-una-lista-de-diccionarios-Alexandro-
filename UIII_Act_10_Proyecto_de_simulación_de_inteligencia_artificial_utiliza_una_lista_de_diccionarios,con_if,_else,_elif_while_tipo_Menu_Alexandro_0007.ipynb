### UIII Act 10 Proyecto de simulación de inteligencia artificial  utiliza una lista de diccionarios,con if, else, elif  while tipo Menu
### Lista medico, Diccionario Los medicos

```python
directorio_medicos = []
medico = {}
print("--- SISTEMA DE INTELIGENCIA MÉDICA v5.0 ---")

while True:
    print("\n--- MENÚ DE CONTROL ---")
    print("1. Registrar un médico (Entrada de Datos)")
    print("2. Ver lista de personal (Proceso de Memoria)")
    print("3. Consultar a la IA (Lógica de Recomendación)")
    print("4. Apagar sistema")

    opcion = input("Seleccione una opción (1-4): ")

    # CASO 1: Registro de Médicos en la base de datos
    if opcion == "1":
        medico = {}

        # Captura de datos interactiva [7]
        medico["nombre"] = input("IA: ¿Cuál es el nombre del médico?: ")
        medico["especialidad"] = input("IA: Especialidad (ej: general, cirujano, pediatra): ")
        medico["años_exp"] = int(input("IA: ¿Cuántos años de experiencia tiene?: "))

        # Agregamos el diccionario (la ficha) a la lista global [3]
        directorio_medicos.append(medico)
        print("IA: Médico registrado correctamente en mi base de datos.")

    # CASO 2: Visualizar la memoria del sistema
    elif opcion == "2":
        print("\n--- DIRECTORIO MÉDICO ACTUAL ---")
        if directorio_medicos == []:
            print("IA AVISO: No hay datos guardados todavía.")
        else:
            # Usamos un ciclo for para recorrer cada diccionario en la lista [8]
            for m in directorio_medicos:
                print("Nombre: " + m["nombre"] + " | Especialidad: " + m["especialidad"])
                print("Trayectoria: " + str(m["años_exp"]) + " años")
                print("-" * 35)

    # CASO 3: Simulación de Inteligencia Artificial (Toma de Decisiones) [18.8.6]
    elif opcion == "3":
        if directorio_medicos == []:
            print("IA ERROR: Base de datos vacía. Por favor use la opción 1 primero.")
        else:
            print("\nIA: Iniciando motor de lógica deductiva...")
            requerimiento = input("¿Qué especialidad necesita el paciente?: ")
            exp_necesaria = int(input("¿Años mínimos de experiencia requeridos?: "))
            encontrado = False
            # La IA busca en la lista de diccionarios al candidato ideal [9, 10]
            for m in directorio_medicos:
                if m["especialidad"] == requerimiento and m["años_exp"] >= exp_necesaria:
                    print("\n--- RECOMENDACIÓN DE LA IA ---")
                    print("He analizado los perfiles y sugiero al Dr(a). " + m["nombre"])
                    print("Diagnóstico: El perfil cumple con la especialidad y el rango de experiencia.")
                    encontrado = True
                    # Detiene la búsqueda al encontrar la primera coincidencia lógica [11]
                    break

            if encontrado == False:
                print("IA: Tras analizar mi memoria, no detecto médicos para esos criterios.")

    # CASO 4: Salida segura del programa
    elif opcion == "4":
        print("IA: Desconectando circuitos de salud... ¡Tenga un buen día!")
        break
print("Elaborado por Alexandro Aguilar NL = 1 ")
```
